```python
# Implement advanced data augmentation techniques
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras import layers
import albumentations as A
import cv2

print("🎨 Implementing Advanced Data Augmentation Techniques")
print("=" * 55)

# 1. Advanced Albumentations-based augmentation
def get_advanced_albumentations():
    """Create advanced augmentation pipeline using Albumentations"""
    return A.Compose([
        A.Resize(224, 224),
        A.OneOf([
            A.RandomBrightnessContrast(brightness_limit=0.3, contrast_limit=0.3, p=0.8),
            A.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.3, hue=0.1, p=0.8),
            A.HueSaturationValue(hue_shift_limit=20, sat_shift_limit=30, val_shift_limit=20, p=0.8)
        ], p=0.9),
        A.OneOf([
            A.MotionBlur(blur_limit=7, p=0.7),
            A.GaussianBlur(blur_limit=(3, 7), p=0.7),
            A.GaussNoise(var_limit=(10, 50), p=0.7)
        ], p=0.5),
        A.OneOf([
            A.ShiftScaleRotate(shift_limit=0.1, scale_limit=0.2, rotate_limit=30, p=0.8),
            A.ElasticTransform(alpha=120, sigma=120 * 0.05, alpha_affine=120 * 0.03, p=0.7),
            A.OpticalDistortion(distort_limit=0.3, shift_limit=0.1, p=0.7)
        ], p=0.8),
        A.OneOf([
            A.GridDistortion(p=0.6),
            A.Perspective(scale=(0.05, 0.1), p=0.6)
        ], p=0.4),
        A.RandomCrop(height=200, width=200, p=0.5),
        A.Resize(224, 224),
        A.HorizontalFlip(p=0.5),
        A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    ])

# 2. Create CutMix function
def cutmix(batch_x, batch_y, alpha=1.0):
    """Apply CutMix augmentation"""
    batch_size = tf.shape(batch_x)[0]
    image_h, image_w = tf.shape(batch_x)[1], tf.shape(batch_x)[2]
    
    # Sample lambda from Beta distribution
    lam = tf.random.uniform([], 0.0, 1.0)
    
    # Sample random indices for mixing
    indices = tf.random.shuffle(tf.range(batch_size))
    
    # Calculate bbox coordinates
    cut_ratio = tf.sqrt(1 - lam)
    cut_w = tf.cast(image_w * cut_ratio, tf.int32)
    cut_h = tf.cast(image_h * cut_ratio, tf.int32)
    
    # Random center point
    cx = tf.random.uniform([], 0, image_w, dtype=tf.int32)
    cy = tf.random.uniform([], 0, image_h, dtype=tf.int32)
    
    # Calculate bbox
    bbx1 = tf.clip_by_value(cx - cut_w // 2, 0, image_w)
    bby1 = tf.clip_by_value(cy - cut_h // 2, 0, image_h)
    bbx2 = tf.clip_by_value(cx + cut_w // 2, 0, image_w)
    bby2 = tf.clip_by_value(cy + cut_h // 2, 0, image_h)
    
    # Create mixed images
    mixed_x = tf.identity(batch_x)
    shuffled_x = tf.gather(batch_x, indices)
    
    # Apply cutmix to the bbox region
    mixed_x = tf.tensor_scatter_nd_update(
        mixed_x,
        tf.stack([
            tf.repeat(tf.range(batch_size), (bbx2-bbx1)*(bby2-bby1)),
            tf.tile(tf.repeat(tf.range(bby1, bby2), bbx2-bbx1), [batch_size]),
            tf.tile(tf.tile(tf.range(bbx1, bbx2), [bby2-bby1]), [batch_size])
        ], axis=1),
        tf.reshape(shuffled_x[:, bby1:bby2, bbx1:bbx2], [-1, 3])
    )
    
    # Mix labels
    shuffled_y = tf.gather(batch_y, indices)
    lam_adjusted = 1 - ((bbx2 - bbx1) * (bby2 - bby1) / (image_h * image_w))
    mixed_y = lam_adjusted * batch_y + (1 - lam_adjusted) * shuffled_y
    
    return mixed_x, mixed_y

# 3. Create MixUp function
def mixup(batch_x, batch_y, alpha=0.2):
    """Apply MixUp augmentation"""
    batch_size = tf.shape(batch_x)[0]
    
    # Sample mixing coefficient
    lam = tf.random.uniform([], 0.0, 1.0)
    
    # Sample random indices
    indices = tf.random.shuffle(tf.range(batch_size))
    
    # Mix images and labels
    mixed_x = lam * batch_x + (1 - lam) * tf.gather(batch_x, indices)
    mixed_y = lam * batch_y + (1 - lam) * tf.gather(batch_y, indices)
    
    return mixed_x, mixed_y

print("✅ Advanced augmentation functions created:")
print("   - Albumentations pipeline with color, geometric, and noise augmentations")
print("   - CutMix implementation for patch-based mixing")
print("   - MixUp implementation for linear interpolation")

# 4. Test augmentation on sample images
sample_img_path = None
for class_name in valid_classes:
    class_path = f"/tmp/Teeth_Dataset/Training/{class_name}"
    if os.path.exists(class_path):
        images = [f for f in os.listdir(class_path) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
        if images:
            sample_img_path = os.path.join(class_path, images[0])
            break

if sample_img_path:
    # Load and show original image
    original_img = cv2.imread(sample_img_path)
    original_img = cv2.cvtColor(original_img, cv2.COLOR_BGR2RGB)
    
    # Apply augmentations
    aug_pipeline = get_advanced_albumentations()
    
    fig, axes = plt.subplots(2, 4, figsize=(16, 8))
    fig.suptitle('Advanced Data Augmentation Examples', fontsize=16, fontweight='bold')
    
    # Original image
    axes[0, 0].imshow(original_img)
    axes[0, 0].set_title('Original Image', fontweight='bold')
    axes[0, 0].axis('off')
    
    # Generate augmented versions
    aug_titles = [
        'Color Enhancement', 'Motion Blur + Rotation', 'Elastic Transform',
        'Grid Distortion', 'Perspective + Noise', 'Multi-Transform', 'Extreme Augmentation'
    ]
    
    for i in range(7):
        augmented = aug_pipeline(image=original_img)['image']
        
        row, col = (i // 4), (i % 4)
        if i < 4:
            row, col = 0, i + 1
        else:
            row, col = 1, i - 4
            
        axes[row, col].imshow(augmented)
        axes[row, col].set_title(aug_titles[i], fontsize=10)
        axes[row, col].axis('off')
    
    plt.tight_layout()
    plt.show()
    
    print("✅ Augmentation visualization completed")
else:
    print("⚠️ No sample image found for augmentation demonstration")
```

```python
# Build Advanced ResNet-50 Architecture from Scratch
import tensorflow as tf
from tensorflow.keras import layers, Model
from tensorflow.keras.applications import ResNet50
from tensorflow.keras.optimizers import Adam, SGD
from tensorflow.keras.callbacks import ReduceLROnPlateau, EarlyStopping, ModelCheckpoint

print("🏗️ Building Advanced ResNet-50 Architecture for Teeth Classification")
print("=" * 65)

def residual_block(x, filters, stride=1, use_projection=False):
    """Create a residual block with batch normalization and dropout"""
    
    # Save input for skip connection
    shortcut = x
    
    # First convolution
    x = layers.Conv2D(filters, 3, strides=stride, padding='same', use_bias=False)(x)
    x = layers.BatchNormalization()(x)
    x = layers.ReLU()(x)
    x = layers.Dropout(0.1)(x)  # Light dropout between conv layers
    
    # Second convolution
    x = layers.Conv2D(filters, 3, strides=1, padding='same', use_bias=False)(x)
    x = layers.BatchNormalization()(x)
    
    # Projection shortcut if needed
    if use_projection or stride != 1:
        shortcut = layers.Conv2D(filters, 1, strides=stride, padding='same', use_bias=False)(shortcut)
        shortcut = layers.BatchNormalization()(shortcut)
    
    # Add shortcut connection
    x = layers.Add()([x, shortcut])
    x = layers.ReLU()(x)
    
    return x

def bottleneck_block(x, filters, stride=1, use_projection=False):
    """Create a bottleneck residual block for deeper networks"""
    
    shortcut = x
    
    # 1x1 conv to reduce dimensions
    x = layers.Conv2D(filters // 4, 1, strides=1, padding='same', use_bias=False)(x)
    x = layers.BatchNormalization()(x)
    x = layers.ReLU()(x)
    
    # 3x3 conv
    x = layers.Conv2D(filters // 4, 3, strides=stride, padding='same', use_bias=False)(x)
    x = layers.BatchNormalization()(x)
    x = layers.ReLU()(x)
    x = layers.Dropout(0.1)(x)
    
    # 1x1 conv to restore dimensions
    x = layers.Conv2D(filters, 1, strides=1, padding='same', use_bias=False)(x)
    x = layers.BatchNormalization()(x)
    
    # Projection shortcut if needed
    if use_projection or stride != 1:
        shortcut = layers.Conv2D(filters, 1, strides=stride, padding='same', use_bias=False)(shortcut)
        shortcut = layers.BatchNormalization()(shortcut)
    
    # Add shortcut connection
    x = layers.Add()([x, shortcut])
    x = layers.ReLU()(x)
    
    return x

def create_advanced_resnet(input_shape=(224, 224, 3), num_classes=7):
    """Create advanced ResNet-50 architecture with modern improvements"""
    
    inputs = layers.Input(shape=input_shape)
    
    # Stem: Initial convolution and pooling
    x = layers.Conv2D(64, 7, strides=2, padding='same', use_bias=False)(inputs)
    x = layers.BatchNormalization()(x)
    x = layers.ReLU()(x)
    x = layers.MaxPooling2D(3, strides=2, padding='same')(x)
    
    # Stage 1: 3 bottleneck blocks (64 filters)
    x = bottleneck_block(x, 256, stride=1, use_projection=True)
    x = bottleneck_block(x, 256, stride=1)
    x = bottleneck_block(x, 256, stride=1)
    
    # Stage 2: 4 bottleneck blocks (128 filters)
    x = bottleneck_block(x, 512, stride=2, use_projection=True)
    x = bottleneck_block(x, 512, stride=1)
    x = bottleneck_block(x, 512, stride=1)
    x = bottleneck_block(x, 512, stride=1)
    
    # Stage 3: 6 bottleneck blocks (256 filters)
    x = bottleneck_block(x, 1024, stride=2, use_projection=True)
    x = bottleneck_block(x, 1024, stride=1)
    x = bottleneck_block(x, 1024, stride=1)
    x = bottleneck_block(x, 1024, stride=1)
    x = bottleneck_block(x, 1024, stride=1)
    x = bottleneck_block(x, 1024, stride=1)
    
    # Stage 4: 3 bottleneck blocks (512 filters)
    x = bottleneck_block(x, 2048, stride=2, use_projection=True)
    x = bottleneck_block(x, 2048, stride=1)
    x = bottleneck_block(x, 2048, stride=1)
    
    # Global Average Pooling
    x = layers.GlobalAveragePooling2D()(x)
    
    # Dropout for regularization
    x = layers.Dropout(0.5)(x)
    
    # Dense layers for classification
    x = layers.Dense(1024, activation='relu')(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(0.3)(x)
    
    x = layers.Dense(512, activation='relu')(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(0.2)(x)
    
    # Output layer
    outputs = layers.Dense(num_classes, activation='softmax', name='predictions')(x)
    
    model = Model(inputs, outputs, name='Advanced_ResNet50_TeethClassifier')
    return model

# Create the model
print("🔧 Building custom ResNet-50 architecture...")
custom_model = create_advanced_resnet(input_shape=(224, 224, 3), num_classes=7)

print(f"✅ Model created successfully!")
print(f"📊 Total parameters: {custom_model.count_params():,}")

# Display model architecture (abbreviated)
print(f"\n🏗️ Model Architecture Summary:")
print(f"Input Shape: (224, 224, 3)")
print(f"Architecture: Advanced ResNet-50 with bottleneck blocks")
print(f"Output Classes: 7 (CaS, CoS, Gum, MC, OC, OLP, OT)")
print(f"Regularization: Batch Normalization + Dropout (0.1, 0.2, 0.3, 0.5)")
print(f"Activation: ReLU + Softmax output")
```

```python
# Implement Transfer Learning with Pre-trained Models
from tensorflow.keras.applications import ResNet50, EfficientNetB4, InceptionV3
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import os

print("🔄 Implementing Transfer Learning Strategy")
print("=" * 45)

def create_transfer_learning_model(base_model_name='ResNet50', input_shape=(224, 224, 3), num_classes=7):
    """Create transfer learning model with pre-trained weights"""
    
    # Choose base model
    if base_model_name == 'ResNet50':
        base_model = ResNet50(
            weights='imagenet',
            include_top=False,
            input_shape=input_shape
        )
    elif base_model_name == 'EfficientNetB4':
        base_model = EfficientNetB4(
            weights='imagenet',
            include_top=False,
            input_shape=input_shape
        )
    elif base_model_name == 'InceptionV3':
        base_model = InceptionV3(
            weights='imagenet',
            include_top=False,
            input_shape=(299, 299, 3)  # InceptionV3 requires 299x299
        )
    
    # Freeze base model initially
    base_model.trainable = False
    
    # Add custom classification head
    inputs = layers.Input(shape=input_shape)
    
    # Preprocess input for the specific model
    if base_model_name == 'InceptionV3' and input_shape != (299, 299, 3):
        x = layers.Lambda(lambda x: tf.image.resize(x, (299, 299)))(inputs)
    else:
        x = inputs
        
    # Apply base model
    x = base_model(x, training=False)
    
    # Add custom head
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dropout(0.5)(x)
    
    # Dense layers with batch normalization
    x = layers.Dense(1024, activation='relu')(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(0.3)(x)
    
    x = layers.Dense(512, activation='relu')(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(0.2)(x)
    
    # Output layer
    outputs = layers.Dense(num_classes, activation='softmax')(x)
    
    model = Model(inputs, outputs, name=f'TransferLearning_{base_model_name}')
    return model, base_model

# Create multiple transfer learning models
models_dict = {}

print("🏗️ Creating Transfer Learning Models:")

# 1. ResNet50 Transfer Learning
print("   📦 Building ResNet50 transfer model...")
resnet_model, resnet_base = create_transfer_learning_model('ResNet50')
models_dict['ResNet50'] = {'model': resnet_model, 'base': resnet_base}
print(f"      ✅ ResNet50: {resnet_model.count_params():,} parameters")

# 2. EfficientNetB4 Transfer Learning  
print("   📦 Building EfficientNetB4 transfer model...")
efficient_model, efficient_base = create_transfer_learning_model('EfficientNetB4')
models_dict['EfficientNetB4'] = {'model': efficient_model, 'base': efficient_base}
print(f"      ✅ EfficientNetB4: {efficient_model.count_params():,} parameters")

print(f"\n🎯 Transfer Learning Strategy:")
print(f"   Phase 1: Train with frozen base (fine-tune head only)")
print(f"   Phase 2: Unfreeze top layers and fine-tune end-to-end")
print(f"   Phase 3: Ensemble multiple models for final predictions")

# Define gradual unfreezing strategy
def unfreeze_model_gradually(model, base_model, unfreeze_percentage=0.3):
    """Gradually unfreeze layers in the base model"""
    
    total_layers = len(base_model.layers)
    layers_to_unfreeze = int(total_layers * unfreeze_percentage)
    
    # Unfreeze from the end (top layers first)
    for layer in base_model.layers[-layers_to_unfreeze:]:
        layer.trainable = True
    
    print(f"   🔓 Unfrozen {layers_to_unfreeze}/{total_layers} layers ({unfreeze_percentage*100}%)")
    
    return model

print(f"\n✅ Transfer learning models created successfully!")
print(f"   - Models ready for phase-based training")
print(f"   - Gradual unfreezing strategy implemented")
print(f"   - Ensemble capability prepared")
```
