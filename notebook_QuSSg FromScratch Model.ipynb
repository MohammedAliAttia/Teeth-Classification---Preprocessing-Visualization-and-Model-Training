```python
# Analyze dataset structure and class distribution
from collections import defaultdict

print("📊 PHASE 1: DATASET ANALYSIS & OPTIMIZATION")
print("=" * 50)

data_path = "/tmp/Teeth_Dataset"
class_counts = defaultdict(lambda: defaultdict(int))
class_names = set()

# Count images in each class for each split
for split in ['Training', 'Validation', 'Testing']:
    split_path = os.path.join(data_path, split)
    if os.path.exists(split_path):
        for class_name in os.listdir(split_path):
            class_path = os.path.join(split_path, class_name)
            if os.path.isdir(class_path):
                # Count images (jpg, jpeg, png)
                image_count = len([f for f in os.listdir(class_path) 
                                 if f.lower().endswith(('.jpg', '.jpeg', '.png'))])
                if image_count > 0:  # Only count classes with images
                    class_counts[split][class_name] = image_count
                    class_names.add(class_name)

# Clean class names (remove any non-class directories)
valid_classes = sorted([cls for cls in class_names if not cls.startswith('output')])

print(f"🎯 Dataset Overview:")
print(f"   📊 Number of classes: {len(valid_classes)}")
print(f"   🏷️ Classes: {valid_classes}")

# Create comprehensive analysis
summary_data = []
for class_name in valid_classes:
    train_count = class_counts['Training'].get(class_name, 0)
    val_count = class_counts['Validation'].get(class_name, 0)
    test_count = class_counts['Testing'].get(class_name, 0)
    total = train_count + val_count + test_count
    
    if total > 0:  # Only include classes with actual data
        summary_data.append({
            'Class': class_name,
            'Training': train_count,
            'Validation': val_count, 
            'Testing': test_count,
            'Total': total,
            'Train_Ratio': train_count / total if total > 0 else 0
        })

df_analysis = pd.DataFrame(summary_data)
print(f"\n📈 Class Distribution Analysis:")
print(df_analysis.to_string(index=False))

# Calculate key metrics
total_images = df_analysis['Total'].sum()
min_class = df_analysis.loc[df_analysis['Total'].idxmin()]
max_class = df_analysis.loc[df_analysis['Total'].idxmax()]
imbalance_ratio = max_class['Total'] / min_class['Total'] if min_class['Total'] > 0 else 0

print(f"\n🔍 Critical Analysis:")
print(f"   📊 Total images: {total_images}")
print(f"   ⚖️ Class imbalance ratio: {imbalance_ratio:.2f}x")
print(f"   📉 Smallest class: {min_class['Class']} ({min_class['Total']} images)")
print(f"   📈 Largest class: {max_class['Class']} ({max_class['Total']} images)")

if imbalance_ratio > 2.0:
    print(f"   ⚠️ Significant imbalance detected - will use class weighting")
else:
    print(f"   ✅ Reasonable class balance")
```

```python
# Implement Advanced Data Augmentation Pipeline for Dental Images
print("🎨 IMPLEMENTING ADVANCED DATA AUGMENTATION PIPELINE")
print("=" * 55)

class AdvancedDentalAugmentation:
    """
    Specialized data augmentation for dental/teeth images
    Designed to increase dataset diversity while preserving anatomical structure
    """
    
    def __init__(self, target_size=(224, 224)):
        self.target_size = target_size
        
    def create_advanced_generator(self, rescale_factor=1./255):
        """Create advanced data generator with dental-specific augmentations"""
        
        return ImageDataGenerator(
            rescale=rescale_factor,
            
            # Geometric transformations (conservative for medical images)
            rotation_range=25,           # Moderate rotation
            width_shift_range=0.15,      # Horizontal shifts
            height_shift_range=0.15,     # Vertical shifts
            shear_range=0.1,            # Slight shearing
            zoom_range=0.2,             # Zoom in/out
            
            # Brightness and contrast (important for X-rays/dental photos)
            brightness_range=[0.8, 1.3], # Brightness variation
            channel_shift_range=0.1,     # Color channel shifts
            
            # Flipping
            horizontal_flip=True,        # Mirror images (dental symmetry)
            vertical_flip=False,         # Don't flip vertically (teeth don't work upside down)
            
            # Fill mode for transformations
            fill_mode='nearest',
            
            # Validation split for training
            validation_split=0.0,       # We have separate validation set
            
            # Preprocessing function for additional augmentations
            preprocessing_function=self.custom_preprocessing
        )
    
    def custom_preprocessing(self, img):
        """Custom preprocessing function for additional augmentations"""
        import numpy as np
        
        # Random noise addition (simulate real-world conditions)
        if np.random.random() > 0.7:
            noise = np.random.normal(0, 0.01, img.shape)
            img = np.clip(img + noise, 0, 1)
        
        # Random contrast adjustment
        if np.random.random() > 0.6:
            factor = np.random.uniform(0.8, 1.2)
            img = np.clip(img * factor, 0, 1)
        
        # Random gamma correction (lighting conditions)
        if np.random.random() > 0.5:
            gamma = np.random.uniform(0.8, 1.2)
            img = np.power(img, gamma)
        
        return img
    
    def create_validation_generator(self, rescale_factor=1./255):
        """Create validation generator with minimal augmentation"""
        return ImageDataGenerator(
            rescale=rescale_factor,
            # Only rescaling for validation - no augmentation
        )

# Initialize augmentation system
print("🔧 Setting up advanced data augmentation...")
augmentation = AdvancedDentalAugmentation(target_size=(224, 224))

# Create data generators
train_datagen = augmentation.create_advanced_generator()
val_datagen = augmentation.create_validation_generator()
test_datagen = augmentation.create_validation_generator()

print("✅ Advanced augmentation generators created!")

# Setup data flow from directories
batch_size = 16  # Smaller batch size for better convergence with custom architecture
target_size = (224, 224)

print(f"\n📂 Creating data flows...")
print(f"   🎯 Target size: {target_size}")
print(f"   📦 Batch size: {batch_size}")

# Training generator with heavy augmentation
train_generator = train_datagen.flow_from_directory(
    os.path.join(data_path, 'Training'),
    target_size=target_size,
    batch_size=batch_size,
    class_mode='categorical',
    classes=main_classes,  # Ensure consistent class order
    shuffle=True,
    seed=42
)

# Validation generator with no augmentation
val_generator = val_datagen.flow_from_directory(
    os.path.join(data_path, 'Validation'),
    target_size=target_size,
    batch_size=batch_size,
    class_mode='categorical',
    classes=main_classes,
    shuffle=False,
    seed=42
)

# Test generator with no augmentation
test_generator = test_datagen.flow_from_directory(
    os.path.join(data_path, 'Testing'),
    target_size=target_size,
    batch_size=batch_size,
    class_mode='categorical',
    classes=main_classes,
    shuffle=False,
    seed=42
)

print(f"✅ Data generators configured:")
print(f"   📈 Training samples: {train_generator.samples}")
print(f"   📊 Validation samples: {val_generator.samples}")
print(f"   🧪 Test samples: {test_generator.samples}")

# Verify class mappings
print(f"\n🏷️ Class mappings verification:")
for idx, class_name in enumerate(main_classes):
    gen_idx = train_generator.class_indices[class_name]
    status = "✅" if idx == gen_idx else "❌"
    print(f"   {class_name}: Expected={idx}, Generator={gen_idx} {status}")

# Calculate class weights for imbalanced dataset
from sklearn.utils.class_weight import compute_class_weight

# Get class distribution from training data
class_counts = [class_counts['Training'][cls] for cls in main_classes]
class_weights_array = compute_class_weight(
    'balanced',
    classes=np.arange(len(main_classes)),
    y=np.repeat(np.arange(len(main_classes)), class_counts)
)

# Convert to dictionary format for Keras
class_weight_dict = {i: weight for i, weight in enumerate(class_weights_array)}

print(f"\n⚖️ Calculated class weights for balanced training:")
for idx, (class_name, weight) in enumerate(zip(main_classes, class_weights_array)):
    print(f"   {class_name}: {weight:.3f}")

# Display augmentation examples
print(f"\n🎨 Generating augmentation examples...")

# Get a sample batch
sample_batch_x, sample_batch_y = next(train_generator)
num_examples = 8

fig, axes = plt.subplots(2, 4, figsize=(16, 8))
fig.suptitle('Advanced Data Augmentation Examples for Dental Images', fontsize=16, fontweight='bold')

for i in range(num_examples):
    if i < len(sample_batch_x):
        row, col = i // 4, i % 4
        
        # Display augmented image
        img = sample_batch_x[i]
        axes[row, col].imshow(img)
        
        # Get class label
        class_idx = np.argmax(sample_batch_y[i])
        class_name = main_classes[class_idx]
        
        axes[row, col].set_title(f'{class_name}', fontsize=12, fontweight='bold')
        axes[row, col].axis('off')

plt.tight_layout()
plt.show()

print(f"✅ Augmentation examples displayed")
print(f"🎯 Ready for training with advanced augmentation pipeline!")
```
